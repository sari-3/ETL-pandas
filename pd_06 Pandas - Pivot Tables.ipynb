The last section will be pivot tables! They are one of the most popular statistical tools in the business world which is very popular with tabular data.

The distinction between groupbys and pivot tables can feel muddy however pivot tables contain more dimensionality in their ability to cut and combine different columns. They're very versatile so I highly recommend looking up how people use them ([article](https://towardsdatascience.com/data-science-meet-pivot-tables-a55db867225c)).

Pandas conveniently have their very own pivot table method so it's very easy to apply to a data frame.

## Simple Example

```python
data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B'],
    'Colour': ['Red', 'Blue', 'Red', 'Green', 'Blue', 'Green'],
    'Value': [10, 15, 5, 8, 12, 3]
}

df1 = pd.DataFrame(data)
df1.head()
```

We want to know how many by `colour` and `category`

We could `groupby` `category` and `colour`

However this may not be the best if we want to compare if category affects the number of each colour.

As always if you wish to do something with any method or function and feel lost, the [documentation](https://pandas.pydata.org/docs/reference/api/pandas.pivot_table.html) and [source code](https://github.com/pandas-dev/pandas/blob/v1.3.2/pandas/core/reshape/pivot.py#L54-L108) (source code found in the documentation) should almost always be your first point of contact. This is because it is the most efficient way to see what's available to you, and in time proficiency in consulting these will be invaluable.
