### 📖 Data Dictionary

| Column | Description |
|--------|-------------|
| `track_id` | Unique Spotify track ID |
| `track_name` | Name of the track |
| `track_artist` | Name of the artist |
| `track_popularity` | Popularity score (0-100) |
| `track_album_name` | Name of the album |
| `track_album_release_date` | Album release date |
| `playlist_name` | Name of the playlist |
| `playlist_genre` | Genre of the playlist |
| `playlist_subgenre` | Subgenre of the playlist |
| `danceability` | How suitable for dancing (0.0 - 1.0) |
| `energy` | Intensity and activity measure (0.0 - 1.0) |
| `key` | Musical key (0-11, where 0=C, 1=C#, etc.) |
| `loudness` | Overall loudness in dB |
| `mode` | Major (1) or Minor (0) |
| `speechiness` | Presence of spoken words (0.0 - 1.0) |
| `acousticness` | Confidence of acoustic sound (0.0 - 1.0) |
| `instrumentalness` | Predicts if track has no vocals (0.0 - 1.0) |
| `liveness` | Presence of audience (0.0 - 1.0) |
| `valence` | Musical positiveness (0.0 - 1.0) |
| `tempo` | Beats per minute (BPM) |
| `duration_ms` | Track duration in milliseconds |
