- `brewery_id` — ID linking to the brewery
- `ounces` — Serving size

**Breweries** (`breweries.csv`):
- `id` — Unique brewery identifier (renamed to `brewery_id`)
- `name` — Brewery name
- `city` — City location
- `state` — US state abbreviation
