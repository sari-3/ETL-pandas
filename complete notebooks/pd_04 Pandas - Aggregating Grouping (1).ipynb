No output? why is that? <br>
Whilst pandas can group the relative categorical data in the 'sex' column it can't give any meaningful output as the other columns need to also be compressed. <br>
To achieve this, we must apply an aggregation (a whole formed by combining several separate elements) method, such as count or sum. 

````python
# lets look the count
df.groupby('sex').count()

````

## Multiparameter Grouping and Aggregating

### Grouping
Here we will show grouping by multiple columns to make a double indexed data frame.

````python
# lets groupby species and island and display their counts
df.groupby(['species', 'island']).count()
````

With a simple line of code, we can draw many insights such as the Adiele species inhabiting the most islands, Biscoe having the most inhabitants, as well as Dream and Biscoe both containing two species.

Since we applied the count method, if we look at the values across the columns you might say they are the same. However, due to null values, this isn't the case as evidenced by the sex column showing fewer counts. We can also draw conclusions from here such as if a penguin has one measurement it has them all implying they were done in one go.

### Aggregating
As with groupby we can also apply multiple aggregate functions using the .agg() method.

````python
# lets apply sum and mean
# from now on we will ignore nulls but feel free to include them!
num_cols = ['bill_length_mm', 'body_mass_g']    # choose numeric columns, or the text column stops the mean
df.groupby(['species', 'island'])[num_cols].agg(['sum', 'mean'])
````

````python
# we can also apply them to a single column by using a dictionary
# it may be excessive here but hey its doable!
df.groupby('species').agg({'body_mass_g': 'mean', 
                           'flipper_length_mm': ['min', 'max']})
````

## Creating new columns

Next, I'd like to showcase my favourite method for making new columns in data frames.<br>
To do this we will utilise the [transform](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.transform.html) method with the groupby.

let us add a new column that will be the mean bill length for that species.

## Pivot Tables

The last section will be pivot tables! They are one of the most popular statistical tools in the business world which is very popular with tabular data.

The distinction between groupbys and pivot tables can feel muddy however pivot tables contain more dimensionality in their ability to cut and combine different columns. They're very versatile so I highly recommend looking up how people use them ([article](https://towardsdatascience.com/data-science-meet-pivot-tables-a55db867225c)).

Pandas conveniently have their very own pivot table method so it's very easy to apply to a data frame.

````python
# we will look at the body mass with island and species as the axes to show syntax
df.pivot_table('body_mass_g', index='species', columns='island')
````

````python
# this can also utilize aggregate functions and list arguments as preciously shown
df.pivot_table('body_mass_g', index='species', columns='island', aggfunc=['mean', 'count'])
````

````python
# melt
wide = df.pivot_table('body_mass_g', index='species', columns='island').reset_index()
wide.melt(id_vars='species', var_name='island', value_name='avg_body_mass_g').dropna()
````

As always if you wish to do something with any method or function and feel lost, the [documentation](https://pandas.pydata.org/docs/reference/api/pandas.pivot_table.html) and [source code](https://github.com/pandas-dev/pandas/blob/v1.3.2/pandas/core/reshape/pivot.py#L54-L108) (source code found in the documentation) should almost always be your first point of contact. Here I will clean the nulls but I'll leave the code so you can use it in your investigations before the Missing Data Workbook

````python
# Here's a small function I use to check nulls (feel free to use)
def null_vals(dataframe):
    '''function to show both number of nulls and the percentage of nulls in the whole column'''
    null_vals = dataframe.isnull().sum()
    total_cnt = len(dataframe)
    null_vals = pd.DataFrame(null_vals,columns=['null'])
    null_vals['percent'] = round((null_vals['null']/total_cnt)*100,3)
    return null_vals.sort_values('percent', ascending=False)

null_vals(challenge)
````
