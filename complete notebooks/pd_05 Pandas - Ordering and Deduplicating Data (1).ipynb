### Pandas - Ordering & Deduplicating

This notebook will give a brief introduction to some of the Pandas techniques.

```python
import pandas as pd
```

```python
# Create a DataFrame
data = {'Name': ['John', 'Alice', 'Bob', 'Alice', 'Bob'],
        'Age': [25, 30, 27, 22, 27],
        'City': ['New York', 'London', 'Paris', 'London', 'Paris']}
df = pd.DataFrame(data)
df
```

# Ordering Columns

```python
desired_order = ['Name', 'City', 'Age']

# Reindex the DataFrame with the desired column order
df_reordered = df.reindex(columns=desired_order)
df_reordered
```

# Ordering Rows

Some functions on data are effected by the order the entries are in so it's important to know how to sort them. This also makes it more readable if your investigating a certain thing.

To order the rows, we use the `sort_values` method on our dataframe.

```python
df = df.sort_values(by = ['Name', 'Age'], ascending = [True, True])
df
```

By default, `sort_values` doesn't do it `inplace` so you have to either specify using the keyword as in the code above, or define it as a new variable.

# Deduplicating Data

An issue that can appear in your data is duplicate values. These can throw off your data so can be removed.

```python
df.duplicated().sum()
```

```python
df_unique = df.drop_duplicates()
df_unique
```

Similarly to `sort_values`, `drop_duplicates()` isn't done `inplace` by default but can be done with the `inplace` keyword. However, the code here just defines it with a new variable

You can see that the duplicated value of for Bob has been removed however Alice, who is duplicated for Name and City is kept as the age is different. While this is useful, there could be occasions where we want to remove the duplicate `Alice` eg if one of them was an old entry and the other was an updated entry, we may only want the most recent data.

To do this we can give a `subset` for which the values have to be equal to be considered a duplicate row

```python
df.drop_duplicates(subset = ['Name', 'City'])
```

Note: The first entry is the only one that is kept so you should sort the rows of your dataset prior to droping duplicates to ensure the correct one remains
