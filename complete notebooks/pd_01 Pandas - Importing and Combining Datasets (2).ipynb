### 1.2 Combining Datasets

Once you have more than one database, you can combine them in a variety of ways depending on what you want to achieve

### 1.2.1 Concatenation

Concatenation involves simply sticking two dataframes together. 

```python
## These are dataframes on trending youtube videos in different countries
jan = pd.DataFrame({'payment_id': [1, 2, 3], 'customer_id': [1, 2, 1], 'amount': [2.99, 0.99, 5.99]})
feb = pd.DataFrame({'payment_id': [4, 5], 'customer_id': [3, 1], 'amount': [4.99, 2.99]})
jan
```

```python
feb
```

```python
pd.concat([jan, feb]).reset_index()
```

```python
pd.concat([jan, feb], ignore_index= True)
```

```python
pd.concat([jan.assign(month='Jan'), feb.assign(month='Feb')], ignore_index=True)
```

```python
# Now we concatenate them one after the other
pd.concat([jan,feb], axis = 1)
```

### 1.2.2 Merging and joining

Merging and joining are two ways of accomplishing the same task. This is for when two dataframes share columns and cannot simply be stuck together

```python
# Two dataframes on countries by population and area

pop = pd.read_csv('data/populations.csv')
area = pd.read_csv('data/area.csv')
```

```python
pop.head()
```

```python
area.head()
```

```python
pop.columns
```

```python
area.columns
```

We want to show both area and population, joining on country name

```python
# We want the same name for the columns to be joined on

```

```python
pop = pop.rename(columns={'Country or dependent territory': 'Country'})
area = area.rename(columns={'Country / Dependency': 'Country'})
```

```python
# Merge created a new dataframe from the two

combined_df =pd.merge(pop, area, how = "inner", on = "Country", suffixes = ('_pop', '_area'))
combined_df.head()
```
